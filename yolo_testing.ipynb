```python
import torch

if torch.cuda.is_available():
    print(f"GPU disponible: {torch.cuda.get_device_name(0)}")
    device = 0  # Usar la primera GPU
else:
    print("No se detectó GPU, usando CPU")
    device = 'cpu'
```

```python
from ultralytics import YOLO

model = YOLO("yolov9m.pt")

model.info()
```

```python
# Try with a smaller batch size
results = model.train(
    data="vid_full/vid_full/data.yaml", 
    epochs=300, 
    imgsz=640, 
    device=device, 
    batch=32,  # Reduced from 128
    optimizer="Adam",
    lr0=0.0001,
    nbs=128,
    patience=50,
)
```
